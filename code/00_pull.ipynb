# 00. Pull

**Input:** `../data/raw/Charlotin-hallucination_cases.csv` (Damien Charlotin's AI Hallucination Cases Database).
**Does:** loads the raw database, prints shape / columns / missingness, writes a snapshot for the next notebook.
**Output:** `../data/interim/raw.pkl`.

Source: https://www.damiencharlotin.com/hallucinations/

```python
import pandas as pd
RAW = "../data/raw/Charlotin-hallucination_cases.csv"   # relative path, no hardcoding
```

```python
df = pd.read_csv(RAW)
print("rows, cols:", df.shape)
print("\ncolumns:", list(df.columns))
```

```python
miss = (df.isna().mean()*100).round(1).sort_values(ascending=False)
print("percent missing by column:\n", miss.to_string())
```

```python
df.to_pickle("../data/interim/raw.pkl")
print("wrote ../data/interim/raw.pkl")
```
